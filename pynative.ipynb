```python
def check_palindrome(number):
    print("Original number:",number)
    original_str = str(number)
    reverse_str = original_str[::-1]
    if original_str == reverse_str:
        print("palindrome")
    else:
        print("No Plaindrome")
check_palindrome(122)
    
```

```python
def merge_lst(list1,list2):
    result_list = []
    for num in list1:
        if num % 2 != 0:
            result_list.append(num)
    for num in list2:
        if num % 2 == 0:
            result_list.append(num)
    return result_list

list1 = [10,20,35,33,37,45]
list2 = [30,40,56,55,63,91]

merge_lst(list1,list2)
        
```
